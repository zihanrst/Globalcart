```python
import pandas as pd
from pathlib import Path
```

```python
RAW_DATA_PATH = Path("../data/raw")

files = list(RAW_DATA_PATH.glob("*.csv"))

for file in files:
    print(file.name)
```

```python
for file in files:
    df = pd.read_csv(file)

    print("=" * 80)
    print(f"File: {file.name}")
    print(f"Shape: {df.shape}")
    print("\nColumns:")
    print(df.columns.tolist())
    print()
```

```python
for file in files:
    df = pd.read_csv(file)

    print("=" * 80)
    print(f"File: {file.name}")
    print(f"Rows: {len(df):,}")
    print(f"Columns: {len(df.columns)}")
    print(f"Duplicate rows: {df.duplicated().sum():,}")

    missing = df.isnull().sum()
    missing = missing[missing > 0]

    print("\nMissing values:")
    if len(missing) == 0:
        print("None")
    else:
        print(missing)

    print()
```

```python
profile = []

for file in files:
    df = pd.read_csv(file)

    profile.append({
        "table": file.stem,
        "rows": len(df),
        "columns": len(df.columns),
        "duplicate_rows": df.duplicated().sum(),
        "missing_cells": df.isna().sum().sum(),
        "missing_columns": (df.isna().sum() > 0).sum()
    })

profile_df = pd.DataFrame(profile)

profile_df.sort_values("rows", ascending=False)
```

```python
core_tables = [
    "olist_orders_dataset.csv",
    "olist_customers_dataset.csv",
    "olist_order_items_dataset.csv",
    "olist_products_dataset.csv",
    "olist_sellers_dataset.csv"
]

for table in core_tables:
    df = pd.read_csv(RAW_DATA_PATH / table)

    print("=" * 80)
    print(table)
    print(df.columns.tolist())
    display(df.head(3))
```

```python
orders = pd.read_csv(RAW_DATA_PATH / "olist_orders_dataset.csv")
customers = pd.read_csv(RAW_DATA_PATH / "olist_customers_dataset.csv")
order_items = pd.read_csv(RAW_DATA_PATH / "olist_order_items_dataset.csv")
products = pd.read_csv(RAW_DATA_PATH / "olist_products_dataset.csv")
sellers = pd.read_csv(RAW_DATA_PATH / "olist_sellers_dataset.csv")

print(
    "orders:",
    len(orders),
    orders["order_id"].nunique()
)

print(
    "customers:",
    len(customers),
    customers["customer_id"].nunique()
)

print(
    "products:",
    len(products),
    products["product_id"].nunique()
)

print(
    "sellers:",
    len(sellers),
    sellers["seller_id"].nunique()
)

print(
    "order_items:",
    len(order_items),
    order_items[["order_id", "order_item_id"]].drop_duplicates().shape[0]
)
```

```python
print("Unique customers (customer_id):",
      customers["customer_id"].nunique())

print("Unique real customers (customer_unique_id):",
      customers["customer_unique_id"].nunique())

print("Unique orders:",
      orders["order_id"].nunique())

print("Unique orders in order_items:",
      order_items["order_id"].nunique())

print("Unique products sold:",
      order_items["product_id"].nunique())

print("Unique sellers active:",
      order_items["seller_id"].nunique())
```

```python
print(
    "Orders without customer:",
    (~orders["customer_id"].isin(customers["customer_id"])).sum()
)

print(
    "Order items without order:",
    (~order_items["order_id"].isin(orders["order_id"])).sum()
)

print(
    "Order items without product:",
    (~order_items["product_id"].isin(products["product_id"])).sum()
)

print(
    "Order items without seller:",
    (~order_items["seller_id"].isin(sellers["seller_id"])).sum()
)
```

## Data Profiling Summary

- 99,441 orders and 96,096 unique customers were identified.
- All core primary-key candidates are unique.
- No orphan foreign-key records were found across the five core tables.
- 775 orders have no associated order items, primarily unavailable or canceled orders.
- Missing order timestamps are largely explained by order lifecycle status and should not be removed blindly.
- 610 products lack category/content metadata, while only 2 products lack physical dimensions.
- Raw data will remain unchanged; cleaning outputs will be stored under `data/processed/`.
